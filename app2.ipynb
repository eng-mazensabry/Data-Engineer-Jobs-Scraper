```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.options import Options
import time
import pandas as pd
 
options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)
##options.add_argument("--headless=new")
options.add_argument("--window-size=1920,1080")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
 
 
driver = webdriver.Chrome(options=options)
driver.get('https://wuzzuf.net/jobs/egypt')
time.sleep(5)
driver.refresh()
driver.find_element(By.NAME, 'q').send_keys('Data Engineer')

driver.find_element(By.XPATH,"//button[normalize-space()='Search Jobs']").click()

time.sleep(5)
driver.refresh()
time.sleep(15)
driver.refresh()
list_jobs = []
for page in range(3):
  print (page+1)
  jobs = driver.find_elements(By.CLASS_NAME, 'css-pkv5jc')
  
  
  for job in jobs:
    job_title = job.find_element(By.CLASS_NAME, 'css-o171kl').text
    job_company = job.find_element(By.CLASS_NAME, 'css-ipsyv7').text
    job_location = job.find_element(By.CLASS_NAME, 'css-16x61xq').text
    job_link = job.find_element(By.CLASS_NAME, 'css-o171kl').get_attribute('href')
    tags =  job.find_element(By.CLASS_NAME,'css-5jhz9n').text
    try:
     job_exp = job.find_element(By.CSS_SELECTOR, "div:nth-child(1) > div:nth-child(3) > div:nth-child(2) > span:nth-child(2)").text
    except:
     job_exp = "Not specified"
    list_jobs.append(
        {'job_title': job_title,
         'job_company': job_company,
         'job_location': job_location,
         'job_link': job_link,
         'tags': tags,
         'job_exp': job_exp
         }
    )
  driver. find_element(By.CSS_SELECTOR, "button[class='css-kefogq ezfki8j0']").click()
  time.sleep(5)
df = pd.DataFrame(list_jobs)
df.to_csv('wuzzuf_jobs.csv', index=False)        
```
